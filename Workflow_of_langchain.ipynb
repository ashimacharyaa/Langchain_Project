```python
class Claim(BaseModel): 
    statement:str

    verdict: Literal[
    "Supported",
    "Not Supported",
    "Unclear"
]
    source_id: list[str]=Field(default_factory=list)

    explaination:str
```

```python
c = Claim(
    statement = "Down syndrom involves extra chromosome 21",
    verdict="Supported",
    source_id=["S1"],
    explaination="Source S1 directly suport st"
)

print(c.model_dump())
```

```python
class Student(BaseModel):
    statement: str

    verdict: Literal[
    "Support",
    "Not Support",
    "Dont Know"
]
    source_id: list[str] = Field(default_factory=list)

    explaination:str
```

```python
cd = Student(
    statement ="Down syndrom involves extra chromosome 21",
    verdict="Support",
    source_id=["S1"],
    explaination="Source S1 directly suport st"
)
print(cd.model_dump())
```

```python
class Verification(BaseModel):
    claims: list[Student]
```

```python
c = Verification(
    claims =[
        Student(
            statement ="Down syndrom involves extra chromosome 21",
            verdict="Support",
            source_id=["S1"],
            explaination="Source S1 directly suport st"
        )
    ]
)       
```

```python
print(c)
```

```python
print(c.claims[0].verdict)
```

```python
print(c.claims[0].statement)
```
